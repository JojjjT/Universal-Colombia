```python
# ============================================================
# TAREA 1 — Inserción inicial en MongoDB
# ============================================================

def insertar_album(nombre_coleccion, canciones):
    """Inserta todas las canciones de un álbum en su colección."""
    coleccion = db[nombre_coleccion]
    coleccion.drop()  # Reiniciar si ya existe (idempotente)
    resultado = coleccion.insert_many(canciones)
    return len(resultado.inserted_ids)

console.print("[bold yellow]🚀 Insertando álbumes en MongoDB...[/bold yellow]\n")

for nombre_col, datos in albumes.items():
    n = insertar_album(nombre_col, datos["canciones"])
    console.print(f"[cyan]📀 {datos['autor']} — {datos['titulo_album']}[/cyan]: [green]{n} canciones insertadas en '{nombre_col}'[/green]")

console.print("\n[bold green]✅ Tarea 1 completada — Base de datos inicial creada.[/bold green]")
```

```python
# ============================================================
# VERIFICACIÓN — Tarea 1
# ============================================================

def mostrar_coleccion(nombre_coleccion, autor, album):
    """Muestra el contenido de una colección en tabla Rich."""
    coleccion = db[nombre_coleccion]
    tabla = Table(title=f"🎵 {autor} — {album}", style="cyan")
    tabla.add_column("_id", style="dim", no_wrap=True)
    tabla.add_column("Título", style="bold white")
    tabla.add_column("Año", justify="center")
    tabla.add_column("Género")
    tabla.add_column("Pista", justify="center")
    tabla.add_column("Popularidad", justify="center", style="green")

    for doc in coleccion.find():
        tabla.add_row(
            str(doc.get("_id", "")),
            doc.get("titulo", ""),
            str(doc.get("anio_salida", "")),
            doc.get("genero", ""),
            str(doc.get("numero_pista", "")),
            str(doc.get("popularidad", ""))
        )
    console.print(tabla)

for nombre_col, datos in albumes.items():
    mostrar_coleccion(nombre_col, datos["autor"], datos["titulo_album"])
    print()
```

```python
# ============================================================
# EXPORTACIÓN CSV — Tarea 1
# ============================================================

def exportar_csv(nombre_archivo):
    """Exporta TODA la base de datos a un único CSV."""
    campos = ["_id", "titulo", "anio_salida", "autor", "id_imagen_portada",
              "genero", "duracion_seg", "numero_pista", "popularidad", "coleccion"]

    with open(nombre_archivo, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=campos, extrasaction="ignore")
        writer.writeheader()

        for nombre_col, datos in albumes.items():
            coleccion = db[nombre_col]
            for doc in coleccion.find():
                doc["coleccion"] = nombre_col
                doc["_id"] = str(doc["_id"])
                writer.writerow({k: doc.get(k, "") for k in campos})

    console.print(f"[bold green]✅ CSV exportado: [yellow]{nombre_archivo}[/yellow][/bold green]")

exportar_csv("disquera_tarea1.csv")

# Mostrar primeras filas
with open("disquera_tarea1.csv", encoding="utf-8") as f:
    for i, linea in enumerate(f):
        if i < 8:
            print(linea.strip())
```

```python
# ============================================================
# EXPORTACIÓN BSON (JSON legible) — Tarea 1
# ============================================================
# MongoDB usa BSON nativamente; desde Python exportamos como JSON
# Para BSON binario real, usar: mongodump --uri=<URI> --db BD_DisqueraUniversal

import os
os.makedirs("bson_tarea1", exist_ok=True)

def exportar_bson_json(carpeta):
    """Exporta cada colección como JSON (representación legible de BSON)."""
    for nombre_col, datos in albumes.items():
        coleccion = db[nombre_col]
        documentos = []
        for doc in coleccion.find():
            doc["_id"] = str(doc["_id"])
            documentos.append(doc)

        # Archivo de colección completa
        ruta_col = os.path.join(carpeta, f"{nombre_col}.json")
        with open(ruta_col, "w", encoding="utf-8") as f:
            json.dump(documentos, f, ensure_ascii=False, indent=2)

        # Archivos individuales por documento
        for doc in documentos:
            ruta_doc = os.path.join(carpeta, f"{doc['_id']}.json")
            with open(ruta_doc, "w", encoding="utf-8") as f:
                json.dump(doc, f, ensure_ascii=False, indent=2)

        console.print(f"[cyan]📦 {nombre_col}[/cyan]: {len(documentos)} docs exportados")

exportar_bson_json("bson_tarea1")
console.print(f"[bold green]✅ BSON (JSON) exportados en carpeta bson_tarea1/[/bold green]")
console.print("[dim]Tip: Para BSON binario real → mongodump --uri=<URI> --db BD_DisqueraUniversal[/dim]")
```

```python
# ============================================================
# TAREA 2 — Inserción de canciones adicionales
# ============================================================

def actualizar_album(nombre_coleccion, nuevas_canciones):
    """Inserta nuevas canciones en la colección y verifica integridad."""
    coleccion = db[nombre_coleccion]
    antes = coleccion.count_documents({})

    # Insertar solo si el _id no existe (upsert manual)
    insertados = 0
    for cancion in nuevas_canciones:
        if not coleccion.find_one({"_id": cancion["_id"]}):
            coleccion.insert_one(cancion)
            insertados += 1

    despues = coleccion.count_documents({})
    return antes, insertados, despues

console.print("[bold yellow]🔄 Actualizando álbumes con canciones adicionales...[/bold yellow]\n")

for nombre_col, nuevas in canciones_adicionales.items():
    autor = albumes[nombre_col]["autor"]
    album = albumes[nombre_col]["titulo_album"]
    antes, ins, despues = actualizar_album(nombre_col, nuevas)
    console.print(
        f"[cyan]🎵 {autor} — {album}[/cyan]: "
        f"[dim]{antes} canciones[/dim] → [green]+{ins} nuevas[/green] → "
        f"[bold]{despues} total[/bold]"
    )

console.print("\n[bold green]✅ Tarea 2 completada — Álbumes actualizados.[/bold green]")
```

```python
# ============================================================
# VERIFICACIÓN DE INTEGRIDAD — Tarea 2
# ============================================================

console.print("[bold yellow]🔍 Verificación de integridad post-actualización:[/bold yellow]\n")

tabla_resumen = Table(title="Resumen de Colecciones — Post Tarea 2", style="green")
tabla_resumen.add_column("Colección", style="cyan")
tabla_resumen.add_column("Artista")
tabla_resumen.add_column("Álbum")
tabla_resumen.add_column("Total Canciones", justify="center", style="bold yellow")

for nombre_col, datos in albumes.items():
    n = db[nombre_col].count_documents({})
    tabla_resumen.add_row(nombre_col, datos["autor"], datos["titulo_album"], str(n))

console.print(tabla_resumen)
```

```python
# ============================================================
# EXPORTACIÓN CSV ACTUALIZADO — Tarea 2
# ============================================================

os.makedirs("bson_tarea2", exist_ok=True)

def exportar_csv_completo(nombre_archivo):
    """Exporta el estado actual completo de todas las colecciones."""
    campos = ["_id", "titulo", "anio_salida", "autor", "id_imagen_portada",
              "genero", "duracion_seg", "numero_pista", "popularidad", "coleccion"]

    with open(nombre_archivo, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=campos, extrasaction="ignore")
        writer.writeheader()
        for nombre_col in albumes.keys():
            for doc in db[nombre_col].find():
                doc["coleccion"] = nombre_col
                doc["_id"] = str(doc["_id"])
                writer.writerow({k: doc.get(k, "") for k in campos})

    console.print(f"[bold green]✅ CSV exportado: [yellow]{nombre_archivo}[/yellow][/bold green]")

exportar_csv_completo("disquera_tarea2.csv")

# BSON actualizado
def exportar_bson_actualizado(carpeta):
    for nombre_col in albumes.keys():
        coleccion = db[nombre_col]
        documentos = [{**doc, "_id": str(doc["_id"])} for doc in coleccion.find()]
        with open(os.path.join(carpeta, f"{nombre_col}.json"), "w", encoding="utf-8") as f:
            json.dump(documentos, f, ensure_ascii=False, indent=2)
        for doc in documentos:
            with open(os.path.join(carpeta, f"{doc['_id']}.json"), "w", encoding="utf-8") as f:
                json.dump(doc, f, ensure_ascii=False, indent=2)
    console.print(f"[bold green]✅ BSON actualizados en: [yellow]{carpeta}/[/yellow][/bold green]")

exportar_bson_actualizado("bson_tarea2")
```
